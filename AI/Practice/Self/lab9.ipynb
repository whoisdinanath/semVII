```python
class FamilyRelations:
    def __init__(self):
        self.facts = set()

    def add_fact(self, relation, person1, person2):
        if relation == 'parent':
            self.facts.add(('parent', person1, person2))
        elif relation == 'child':
            self.facts.add(('parent', person2, person1))  # child A, B = B is parent of A

    def get_parents(self, person):
        return {p for (rel, p, c) in self.facts if rel == 'parent' and c == person}

    def get_children(self, person):
        return {c for (rel, p, c) in self.facts if rel == 'parent' and p == person}

    def get_siblings(self, person):
        parents = self.get_parents(person)
        siblings = set()
        for parent in parents:
            siblings.update(self.get_children(parent))
        siblings.discard(person)
        return siblings

    def get_grandparents(self, person):
        grandparents = set()
        for parent in self.get_parents(person):
            grandparents.update(self.get_parents(parent))
        return grandparents

    def get_grandchildren(self, person):
        grandchildren = set()
        for child in self.get_children(person):
            grandchildren.update(self.get_children(child))
        return grandchildren

    def get_uncles_or_aunts(self, person):
        uncles_aunts = set()
        for parent in self.get_parents(person):
            uncles_aunts.update(self.get_siblings(parent))
        return uncles_aunts

    def get_nephews_or_nieces(self, person):
        nephews_nieces = set()
        for sibling in self.get_siblings(person):
            nephews_nieces.update(self.get_children(sibling))
        return nephews_nieces

    def show(self):
        for fact in sorted(self.facts):
            print(f"{fact[0].capitalize()}: {fact[1]} → {fact[2]}")

    def list_relation(self, relation, person):
        relation_map = {
            'parent': self.get_parents,
            'child': self.get_children,
            'sibling': self.get_siblings,
            'grandparent': self.get_grandparents,
            'grandchild': self.get_grandchildren,
            'uncle_or_aunt': self.get_uncles_or_aunts,
            'nephew_or_niece': self.get_nephews_or_nieces
        }

        if relation not in relation_map:
            print(f"Unsupported relation: {relation}")
            return

        related = relation_map[relation](person)
        if related:
            print(f"{relation.capitalize()} of {person}: {', '.join(related)}")
        else:
            print(f"No {relation} found for {person}")

    def query(self, relation, person1, person2):
        relation_map = {
            'parent': self.get_children,
            'child': self.get_parents,
            'sibling': self.get_siblings,
            'grandparent': self.get_grandchildren,
            'grandchild': self.get_grandparents,
            'uncle_or_aunt': self.get_nephews_or_nieces,
            'nephew_or_niece': self.get_uncles_or_aunts
        }

        if relation not in relation_map:
            print(f"Unsupported relation: {relation}")
            return

        related = relation_map[relation](person1)
        result = person2 in related
        print(f"{'Yes' if result else 'No'} — {person1} is {relation} of {person2}")
```

```python
fam = FamilyRelations()
```

```python
# Adding sample facts
fam.add_fact('parent', 'alice', 'bob')
fam.add_fact('parent', 'alice', 'charles')
fam.add_fact('parent', 'bob', 'david')
fam.add_fact('parent', 'charles', 'emma')
```

```python
# Show facts
print("All stored facts:")
fam.show()
```

```python
# List examples
print("\nList:")
fam.list_relation('child', 'alice')
fam.list_relation('sibling', 'bob')
fam.list_relation('grandparent', 'david')
fam.list_relation('uncle_or_aunt', 'david')
fam.list_relation('nephew_or_niece', 'charles')
```

```python
# Query examples
print("\nQuery:")
fam.query('parent', 'alice', 'bob')
fam.query('sibling', 'bob', 'charles')
fam.query('grandparent', 'alice', 'emma')
fam.query('uncle_or_aunt', 'charles', 'david')
```
