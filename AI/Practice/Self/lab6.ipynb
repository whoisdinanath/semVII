```python
import numpy as np

def to_numeric_grid(maze):
    numeric_grid = np.zeros((len(maze), len(maze[0])), dtype=int)
    for i in range(len(maze)):
        for j in range(len(maze[i])):
            if maze[i][j] == 'S':
                numeric_grid[i][j] = 2 
            elif maze[i][j] == 'G':
                numeric_grid[i][j] = 3  
            elif maze[i][j] == 1:
                numeric_grid[i][j] = 1 
            else:
                numeric_grid[i][j] = 0
    return numeric_grid
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors
from collections import deque
import numpy as np
import time


# add maze randomization
# log path and steps taken
# show total visited nodes and execution time

def search_with_visualization(grid, algorithm='bfs', delay=0.1):
    start_pos = None
    goal_pos = None
    for i in range(grid.shape[0]):
        for j in range(grid.shape[1]):
            if grid[i][j] == 2:
                start_pos = (i, j)
            elif grid[i][j] == 3:
                goal_pos = (i, j)
    
    if start_pos is None or goal_pos is None:
        print("Start or goal position not found!")
        return None
    
    colors = ['white', 'black', 'orange', 'red', 'lightgreen', 'blue']
    cmap = matplotlib.colors.ListedColormap(colors)
    
    viz_grid = grid.copy()
    
    if algorithm == 'bfs':
        frontier = deque([(start_pos, [start_pos])])
    else:  # dfs
        frontier = [(start_pos, [start_pos])]
    
    visited = set()
    visited.add(start_pos)
    
    while frontier:
        if algorithm == 'bfs':
            (r, c), path = frontier.popleft()
        else:  # dfs
            (r, c), path = frontier.pop()
        
        if (r, c) == goal_pos:
            print(f"{algorithm.upper()} found goal!")
            print(f"Path length: {len(path)}")
            print(f"Path: {path}")
            
            for pos in path:
                if pos != start_pos and pos != goal_pos:
                    viz_grid[pos[0], pos[1]] = 5
            
            # Final visualization
            return path
        
        if (r, c) != start_pos and (r, c) != goal_pos:
            viz_grid[r, c] = 4
        
        
        neighbors = get_neighbors(grid, r, c)
        for nr, nc in neighbors:
            if (nr, nc) not in visited:
                visited.add((nr, nc))
                new_path = path + [(nr, nc)]
                
                if algorithm == 'bfs':
                    frontier.append(((nr, nc), new_path))
                else:
                    frontier.append(((nr, nc), new_path))
    
    print(f"{algorithm.upper()} could not find a path!")
    return None

print("Running BFS:")
bfs_path = search_with_visualization(grid, 'bfs', 0.05)

print("\nRunning DFS:")
dfs_path = search_with_visualization(grid, 'dfs', 0.05)
```
