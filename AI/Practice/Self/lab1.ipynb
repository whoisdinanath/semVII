```python
import torch.nn as nn
import torch
```

```python
layer1 = nn.Linear(10, 16)
layer2 = nn.Linear(16, 8)
layer3 = nn.Linear(8,1)

model = nn.Sequential(layer1, layer2, layer3)
```

### Lab: How are torch.randn numbers generated?

```python
torch.manual_seed(42)
input1 = torch.randn(10)
input1
```

```python
output = model(input1)
output
```

```python
out1 = layer1(input1)
out1
```

```python
out2 = layer2(out1)
out2
```

```python
outp = layer3(out2)
outp
```
