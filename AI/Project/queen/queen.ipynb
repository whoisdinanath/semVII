```python
import random
import time
```

```python
class NQueenSolution:
    def __init__(self, n, population_size=200, mutation_rate=0.01, max_generations=1000, max_stagnant=100):
        self.n = n
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.max_generations = max_generations
        self.max_stagnant = max_stagnant
        self.max_fitness = (n * (n - 1)) // 2
        self.solutions = set()
        self.stagnant_generations = 0
        self.generation = 0


    def generate_individual(self):
        individual = list(range(self.n))
        random.shuffle(individual)
        return tuple(individual)

    def fitness(self, individual):
        conflicts = 0
        for i in range(self.n):
            for j in range(i + 1, self.n):
                if abs(individual[i] - individual[j]) == abs(i - j):
                    conflicts += 1
        return self.max_fitness - conflicts

    def select_parents(self, population, fitnesses):
        return random.choices(population, weights=fitnesses, k=2)

    def crossover(self, parent1, parent2):
        point = random.randint(1, self.n - 1)
        child = parent1[:point] + tuple(x for x in parent2 if x not in parent1[:point])
        return tuple(child)

    def mutate(self, individual):
        if random.random() < self.mutation_rate:
            i, j = random.sample(range(self.n), 2)
            individual = list(individual)
            individual[i], individual[j] = individual[j], individual[i]
            return tuple(individual)
        return individual

    def evolve(self):
        populations = [self.generate_individual() for _i in range(self.population_size)]

        while self.stagnant_generations < self.max_stagnant and self.generation < self.max_generations:
            fitnesses = [self.fitness(ind) for ind in populations]
            new_population = []

            found_new = False
            for i, f in enumerate(fitnesses):
                if f == self.max_fitness and populations[i] not in self.solutions:
                    self.solutions.add(populations[i])
                    found_new = True
                    print(f"Solution found: {populations[i]} at generation {self.generation} with fitness {f}")

            if found_new:
                self.stagnant_generations = 0
            else:
                self.stagnant_generations += 1

            for _ in range(self.population_size):
                parent1, parent2 = self.select_parents(populations, fitnesses)
                child = self.crossover(parent1, parent2)
                child = self.mutate(child)
                new_population.append(child)
            populations = new_population
            self.generation += 1

        print(f"Total solutions found: {len(self.solutions)}")
        return list(self.solutions)
```

```python
ga_all = NQueenSolution(n=8, population_size=2000, max_generations=15000, max_stagnant=50)

start_time = time.time()  # Start timer

solutions = ga_all.evolve()

end_time = time.time()  # End timer

print(f"Execution time: {end_time - start_time:.4f} seconds")
```
