```python
def plot_histogram(image, title):
    r_ch = image[:, :, 0]
    g_ch = image[:, :, 1]
    b_ch = image[:, :, 2]

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 3, 1)
    plt.hist(r_ch.ravel(), bins=256, color='red')
    plt.title(f'{title} - Red Channel Histogram')
    plt.subplot(1, 3, 2)
    plt.hist(g_ch.ravel(), bins=256, color='green')
    plt.title(f'{title} - Green Channel Histogram')
    plt.subplot(1, 3, 3)
    plt.hist(b_ch.ravel(), bins=256, color='blue')
    plt.title(f'{title} - Blue Channel Histogram')
    plt.tight_layout()

    plt.savefig(f"../Reports/images/lab1/{title.lower()}_histogram.png", dpi=300, bbox_inches='tight')
    plt.show()
plot_histogram(bipul_image, 'Bipul')
plot_histogram(bibek_image, 'Bibek')
plot_histogram(krishna_image, 'Krishna')
```

```python
# Filtering the image
def add_noise_gaussian(image, noise_factor=0.1):
    noise = np.random.normal(loc=0, scale=noise_factor, size=image.shape)
    noisy_image = image + noise
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)
    return noisy_image

noisy_bipul = add_noise_gaussian(bipul_image, noise_factor=20)
noisy_bibek = add_noise_gaussian(bibek_image, noise_factor=20)
noisy_krishna = add_noise_gaussian(krishna_image, noise_factor=20)

def add_salt_and_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02):
    noisy_image = np.copy(image)
    total_pixels = image.size
    num_salt = np.ceil(salt_prob * total_pixels)
    num_pepper = np.ceil(pepper_prob * total_pixels)

    # Salt noise
    coords = [np.random.randint(0, i - 1, int(num_salt)) for i in image.shape]
    noisy_image[coords[0], coords[1], :] = 255

    # Pepper noise
    coords = [np.random.randint(0, i - 1, int(num_pepper)) for i in image.shape]
    noisy_image[coords[0], coords[1], :] = 0

    return noisy_image

noisy_bipul_sp = add_salt_and_pepper_noise(bipul_image, salt_prob=0.02, pepper_prob=0.02)
noisy_bibek_sp = add_salt_and_pepper_noise(bibek_image, salt_prob=0.02, pepper_prob=0.02)
noisy_krishna_sp = add_salt_and_pepper_noise(krishna_image, salt_prob=0.02, pepper_prob=0.02)
```

```python
x_sobel = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

y_sobel = np.array([[1, 2, 1],
                    [0, 0, 0],
                    [-1, -2, -1]])

laplacian_kernel = np.array([[0, 1, 0],
                            [1, -4, 1],
                            [0, 1, 0]]) # from the book

gaussian_kernel = np.array([[1, 2, 1],
                            [2, 4, 2],
                            [1, 2, 1]]) / 16.0  # Normalized Gaussian kernel

def convolve2d(image, kernel):
    image_height, image_width = image.shape
    kernel_height, kernel_width = kernel.shape
    pad_height = kernel_height // 2
    pad_width = kernel_width // 2

    output_image = np.zeros((image_height, image_width))

    padded_image = np.pad(image, ((pad_height, pad_height), (pad_width, pad_width)), mode='constant', constant_values=0)

    for i in range(image_height):
        for j in range(image_width):
            patch = padded_image[i:i + kernel_height, j:j + kernel_width]

            output_image[i, j] = np.sum(patch * kernel)
    return output_image

def sobel_filter(image):
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    sobel_x = convolve2d(gray_image, x_sobel)
    sobel_y = convolve2d(gray_image, y_sobel)

    sobel_magnitude = np.sqrt(sobel_x**2 + sobel_y**2)
    sobel_magnitude = np.clip(sobel_magnitude, 0, 255).astype(np.uint8)

    return sobel_magnitude

def laplacian_filter(image):
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    laplacian_result = convolve2d(gray_image, laplacian_kernel)
    laplacian_result = np.clip(laplacian_result, 0, 255).astype(np.uint8)

    return laplacian_result

def gaussian_filter(image):
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gaussian_result = convolve2d(gray_image, gaussian_kernel)
    gaussian_result = np.clip(gaussian_result, 0, 255).astype(np.uint8)

    return gaussian_result
```

```python
bipul_filtered = sobel_filter(noisy_bipul)
bibek_filtered = sobel_filter(noisy_bibek)
krishna_filtered = sobel_filter(noisy_krishna)
plt.subplot(1, 3, 1)
plt.imshow(bipul_filtered, cmap='gray')
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(bibek_filtered, cmap='gray')
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(krishna_filtered, cmap='gray')
plt.axis('off')
plt.savefig("../Reports/images/lab1/sobel_filtered_images.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
def restore_gaussian(image, kernel_size=5, sigma=1.0):
    kernel = cv2.getGaussianKernel(kernel_size, sigma)
    gaussian_kernel = np.outer(kernel, kernel)
    restored_image = cv2.filter2D(image, -1, gaussian_kernel)
    return restored_image


bip_restored = restore_gaussian(bipul_gaussian)
bib_restored = restore_gaussian(bibek_gaussian)
krishna_restored = restore_gaussian(krishna_gaussian)

fig = plt.figure(figsize=(12, 8))
plt.subplot(1, 3, 1)
plt.imshow(bip_restored)
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(bib_restored)
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(krishna_restored)
plt.axis('off')
plt.savefig("../Reports/images/lab1/restored_gaussian_images.png", dpi=300, bbox_inches='tight')
plt.show()
```
